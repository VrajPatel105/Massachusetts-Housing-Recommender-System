# Understanding the features
## `Note : This data is scraped from zillow.com. Respecting the website policies, I have made sure to not to overload the server. Which is why, I made the scraper in a way that it scrapes only 1 property per minute. And it was done parallely, with 5 other tabs.  Gathering this data took days. `


## This is data and project is only for educational purposes.

## Columns:

`url` : url of the property listed on the website

`image_url` : image url of the property

`beds`: Total number of beds

`baths` : total number of baths

`sqft` : total area covered.

`sqft_lot`: total sqft lot. Other two variants of this are : `sqft_lot_cleaned` and `sqft_lot_categories`

`address`: address of the property

`estimated_monthly_payment`: estimated monthly rent for the property

`property_type`: property type includes : `single family`, `multi-family`, `condo` and `townhouse`

`price_per_sqft`: total price per sqft

`year_built`: property built year

`region`: region of property located. for ex : `731-733 Carew St, Springfield, MA 01104` -> region: `Springfield`

`interior_features`: interior features included in that property

`other_rooms`: other rooms available

`appliances`: appliances offered

`utilities_Electric`: electric utilities

`utilities_Sewer`: utilities for sewer and anything that's special

`utilities_Water`: same as above

`utilities_Utilities`: same as above

`parking_total_spaces`: total parking space available

`parking_parking_features`: parking space features

`walk_score`: walking score. This is out of 100 score. ex . (30/100).

`bike_score`: bike score. This is also out of 100 score.

`transit_score`: transit score. This is also out of 100.

`elementary_school_name`: nearby elementary school name : All school information is collected since it will help the recommendation engine as well

`elementary_school_distance`: elementary school distance from property

`middle_school_name`: nearby middle school name

`middle_school_distance`: nearby middle school distance from property

`high_school_name`: nearby high school name

`high_school_distance`: nearby high school distance from property

`flood_risk`: flood risk in that area. This is out of 10 . ex : (6/10)

`fire_risk` : fire risk. out of 10

`wind_risk` : wind risk. out of 10

`air_risk`: air risk. out of 10

`heat_risk`: heat risk. out of 10

`nearby_cities`: top 5 nearby cities. Nearby form the property's region

`property_history`: property past history. Price trend over the year.

`parking_uncovered_spaces`: uncovered parking spaces

`parking_garage_spaces`: total cars that can be parked in the garage.

`price` : total price of the prpoerty. `This feature is the predicting feature. So i have move this feature all the way to the last`

`sqft_lot_category` : sqft lot categorized to 6 categories based on this functino that was applied in the data cleaning phase.

    def categorize_lot_size(sqft):
        if sqft is None:       # Handle None values
            return 'Unknown'
        elif sqft < 2000:
            return 'Ultra Compact'
        elif sqft < 4000:
            return 'Urban Small'
        elif sqft < 7000:
            return 'Urban Medium'
        elif sqft < 12000:
            return 'Suburban Small'
        elif sqft < 25000:
            return 'Suburban Medium'
        elif sqft < 43560:
            return 'Suburban Large'
        else:
            return 'Rural/Estate'

`sqft_lot_cleaned` : This is simply standardizing all the values to sqft. (conveted acres to sqft.)

`zip_code`: Zip code of the property. This was extracted from the address column. This can play a very important role. And this is a object data type column. Since this is considered as a category
