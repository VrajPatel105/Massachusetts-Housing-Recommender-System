```python
# Price by property type
df.boxplot(column='price', by='property_type')

# Price by region (top regions only)
top_regions = df['region'].value_counts().head(10).index
df[df['region'].isin(top_regions)].boxplot(column='price', by='region')

# Price statistics by category
df.groupby('property_type')['price'].describe()
df.groupby('region')['price_per_sqft'].mean().sort_values(ascending=False)
```

```python
# Cross-tabulations
pd.crosstab(df['property_type'], df['region'])
```

```python
target = 'price'

numerical_cols = ['beds', 'baths', 'sqft', 'sqft_lot_cleaned', 'year_built',
                  'price_per_sqft', 'estimated_monthly_payment', 'parking_total_spaces',
                  'walk_score', 'bike_score', 'elementary_school_distance',
                  'middle_school_distance', 'high_school_distance']

categorical_cols = ['property_type', 'region', 'zip_code', 'sqft_lot_caterogy']
```

```python
for col in numerical_cols:
    plt.figure(figsize=(10, 6))

    # Scatter plot
    plt.subplot(1, 2, 1)
    plt.scatter(df[col], df[target], alpha=0.5)
    plt.xlabel(col)
    plt.ylabel(target)
    plt.title(f'{col} vs {target}')

    # Correlation coefficient
    corr = df[col].corr(df[target])
    plt.text(0.05, 0.95, f'Correlation: {corr:.3f}', transform=plt.gca().transAxes)

    # Distribution of the feature
    plt.subplot(1, 2, 2)
    plt.hist(df[col].dropna(), bins=30)
    plt.xlabel(col)
    plt.title(f'Distribution of {col}')

    plt.tight_layout()
    plt.show()

    print(f"\n{col} vs {target}:")
    print(f"Correlation: {corr:.3f}")
    print(f"Strength: {'Strong' if abs(corr) > 0.7 else 'Moderate' if abs(corr) > 0.3 else 'Weak'}")
```

```python
for col in categorical_cols:
    plt.figure(figsize=(12, 6))

    # Box plot
    plt.subplot(1, 2, 1)
    df.boxplot(column=target, by=col, ax=plt.gca())
    plt.xticks(rotation=45)
    plt.title(f'{target} by {col}')

    # Bar plot of means
    plt.subplot(1, 2, 2)
    means = df.groupby(col)[target].mean().sort_values(ascending=False)
    means.plot(kind='bar')
    plt.xticks(rotation=45)
    plt.title(f'Mean {target} by {col}')

    plt.tight_layout()
    plt.show()

    # Summary statistics
    print(f"\n{col} vs {target}:")
    print(df.groupby(col)[target].describe())
```

```python
# Correlation matrix for all numerical variables
corr_matrix = df[numerical_cols + [target]].corr()
plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0)
plt.title('Correlation Matrix')
plt.show()
```

# `Conclusion`

1. The key price driving factors are beds, baths, sqft, price_per_sqft, estimated_monthly_payment

2. The key price non driving cols : sqft_lot_cleaned, year_build (never expected : - ] , walk_score, bike_score, all school distances )

3. Parking space and estimated monthly payment is slightly correlated to beds, baths and sqft

```python
categorical_cols = ['property_type', 'region', 'zip_code', 'sqft_lot_caterogy']
target = 'price'

for col in categorical_cols:
    # Get top 15 categories by frequency
    top_categories = df[col].value_counts().head(15).index
    df_filtered = df[df[col].isin(top_categories)]

    plt.figure(figsize=(15, 10))

    # Box plot
    plt.subplot(2, 2, 1)
    df_filtered.boxplot(column=target, by=col, ax=plt.gca())
    plt.xticks(rotation=45, ha='right')
    plt.title(f'{target} by {col} (Top 15)')
    plt.tight_layout()

    # Mean price by category (sorted)
    plt.subplot(2, 2, 2)
    means = df_filtered.groupby(col)[target].mean().sort_values(ascending=False)
    means.plot(kind='bar', color='skyblue')
    plt.xticks(rotation=45, ha='right')
    plt.title(f'Mean {target} by {col}')
    plt.ylabel('Mean Price')

    # Count of properties per category
    plt.subplot(2, 2, 3)
    counts = df_filtered[col].value_counts()
    counts.plot(kind='bar', color='lightcoral')
    plt.xticks(rotation=45, ha='right')
    plt.title(f'Property Count by {col}')
    plt.ylabel('Count')

    # Price range (min-max) by category
    plt.subplot(2, 2, 4)
    price_range = df_filtered.groupby(col)[target].agg(['min', 'max'])
    price_range.plot(kind='bar', color=['orange', 'green'])
    plt.xticks(rotation=45, ha='right')
    plt.title(f'Price Range by {col}')
    plt.ylabel('Price')
    plt.legend(['Min Price', 'Max Price'])

    plt.tight_layout()
    plt.show()

    # Summary statistics table
    print(f"\n=== {col.upper()} vs {target.upper()} (Top 15) ===")
    summary = df_filtered.groupby(col)[target].agg(['count', 'mean', 'median', 'std']).round(0)
    summary = summary.sort_values('mean', ascending=False)
    print(summary)

    # Price difference from overall mean
    overall_mean = df[target].mean()
    summary['price_premium'] = summary['mean'] - overall_mean
    print(f"\nPrice Premium vs Overall Mean (${overall_mean:,.0f}):")
    print(summary[['mean', 'price_premium']].head(10))
    print("-" * 50)
```

```python
from scipy.stats import chi2_contingency

categorical_price_corr = {}
for col in categorical_cols:
    top_cats = df[col].value_counts().head(10).index
    filtered_df = df[df[col].isin(top_cats)]

    category_means = filtered_df.groupby(col)[target].mean()

    correlation = filtered_df.groupby(col)[target].mean().corr(
        filtered_df.groupby(col)[target].count()
    )
    categorical_price_corr[col] = correlation

for col, corr in categorical_price_corr.items():
    print(f"{col}: {corr:.3f}")
```

```python
# Analyze property_type properly
target = 'price'

# Basic statistics by property type
print("=== PROPERTY TYPE ANALYSIS ===")
property_stats = df.groupby('property_type')[target].agg(['count', 'mean', 'median', 'std']).round(0)
property_stats = property_stats.sort_values('mean', ascending=False)
print(property_stats)

# Calculate price premium vs overall mean
overall_mean = df[target].mean()
property_stats['price_premium'] = property_stats['mean'] - overall_mean
print(f"\nPrice Premium vs Overall Mean (${overall_mean:,.0f}):")
print(property_stats[['mean', 'price_premium']])
```

```python
from scipy import stats

# Test if property types have significantly different prices
property_groups = [group['price'].values for name, group in df.groupby('property_type')]
f_stat, p_value = stats.f_oneway(*property_groups)

print(f"ANOVA F-statistic: {f_stat:.3f}")
print(f"P-value: {p_value:.3f}")
print(f"Significant difference: {'Yes' if p_value < 0.05 else 'No'}")
```

This confirms property type should be a key variable in any pricing model, with single family homes as the reference category commanding the highest values in the dataset.
